```python
import pandas as pd
```

```python
# Lectura y Verificación Estructural
df_urgencias = pd.read_csv("urgencias_hospitales.csv")
```

```python
# Selección Simple y Proyección Múltiple
# serie_nivel_triage = df_urgencias["Nivel_Triage"]
vista_tabular = df_urgencias[["Hospital", "Especialidad", "Tiempo_Espera_Min"]]
vista_tabular
```

```python
# Acceso por Etiqueta de Índice (loc)
df_urgencias.loc[75]
```

```python
# Tratamiento de Redundancias
df_urgencias.drop_duplicates()
```

```python
# Filtrado Booleano Simple
df_urgencias[df_urgencias["Edad_Paciente"] > 65]
```

```python
# Filtrado Categórico por Lista (isin)
df_urgencias[df_urgencias["Hospital"].isin(["Hospital Santa Sofía", "Clínica San Marcel"])]
```

```python
# Filtrado por Intervalo Clínico (between)
df_urgencias[df_urgencias["Tiempo_Espera_Min"].between(30.0, 90.0)]
```

```python
# Consultas Directas con .query()
df_urgencias.query("Nivel_Triage == 'Triage I' and Costo_Procedimiento > 300000")
```

```python
# Filtrado Compuesto con Métodos de Texto
mascara = (df_urgencias["Destino_Alta"] == "Hospitalización") & (df_urgencias["Especialidad"].str.startswith("M"))
mascara
```

```python
# Imputación Estratificada de Faltantes
df_urgencias["Tiempo_Espera_Min"] = df_urgencias["Tiempo_Espera_Min"].fillna(df_urgencias["Tiempo_Espera_Min"].mean())
df_urgencias["Costo_Procedimiento"] = df_urgencias["Costo_Procedimiento"].fillna(0.0)
df_urgencias
```

```python
# Transformación Personalizada de Atributos (apply)
def clasificar(edad):
    if edad < 18:
        return "Pediatrico"
    elif edad <= 60:
        return "Adulta"
    else: 
        return "Adulto Mayor"

df_urgencias["Grupo_Etario"] = df_urgencias["Edad_Paciente"].apply(clasificar)
df_urgencias
```

```python
# Renombrado y Reorganización de Campos
df_urgencias = df_urgencias.rename(columns={
    "Tiempo_Espera_Min":"Minutos_Espera",
    "Destino_Alta":"Egreso"
})

df_urgencias = df_urgencias[["Grupo_Etario"] + [c for c in df_urgencias.columns if c != "Grupo_Etario"]]
df_urgencias
```

```python
# Agregación por Centro Hospitalario
df_urgencias.groupby("Hospital").agg(
    Promedio_Espera=("Minutos_Espera", "mean"),
    Costo_Total=("Costo_Procedimiento", "sum"),
)
```

```python
# Medición de Rango Clínico
def rango(series):
    return series.max() - series.min()

df_urgencias.groupby("Nivel_Triage")[["Minutos_Espera", "Costo_Procedimiento"]].agg(rango)
```

```python
# Agrupación Multinivel Avanzada
df_urgencias.groupby(["Hospital", "Nivel_Triage"]).agg(
    Edad_Promedio=("Edad_Paciente", "mean"),
    Costo_Total=("Costo_Procedimiento", "sum"),
)
```
